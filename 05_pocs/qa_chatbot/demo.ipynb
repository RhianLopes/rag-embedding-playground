# PoC: Q&A Chatbot

Chatbot que responde perguntas sobre documentos locais usando RAG.

## Stack
- sentence-transformers (embeddings)
- Qdrant (vector store)
- Ollama/llama3.2 (LLM)
- Historico de conversa (context window)

**Prerequisito:** `docker compose up -d` e `docker exec ollama ollama pull llama3.2`

```python
import sys
sys.path.insert(0, '../..')
sys.path.insert(0, '.')

from pipeline import QAChatbot
from rich.console import Console

console = Console()
chatbot = QAChatbot()
print('Chatbot pronto!')
```

```python
# Conversa demo
perguntas = [
    'O que e RAG e quais sao suas principais vantagens?',
    'Como o chunking afeta a qualidade do RAG?',
    'Quais metricas do RAGAs devo monitorar?',
]

for pergunta in perguntas:
    print(f'\nPergunta: {pergunta}')
    result = chatbot.chat(pergunta)
    print(f'\nResposta:\n{result["answer"][:500]}')
    print(f'\nFontes: {[r["payload"]["source"] for r in result["sources"][:3]]}')
    print(f'Timing: retrieve={result["timings"]["retrieve_ms"]:.0f}ms | generate={result["timings"]["generate_ms"]:.0f}ms')
    print('='*60)
```

```python
# Testar memoria conversacional
print(f'Historico da conversa: {len(chatbot.conversation_history)} turnos')
print(f'Ultimo tema discutido: {chatbot.conversation_history[-1]["question"][:50]}')

# Pergunta que usa contexto anterior
followup = 'Voce pode elaborar mais sobre a ultima metrica que mencionou?'
result = chatbot.chat(followup)
print(f'\nFollowup: {followup}')
print(f'Resposta: {result["answer"][:400]}')
```
